# 99. Executar tudo

Roda o pipeline na ordem. A coleta de clima fica de fora, porque depende da internet.

```python
import nbformat
from nbclient import NotebookClient

ORDEM = ["02_bronze_silver", "03_gold", "04_analise_exploratoria", "05_torneio_de_algoritmos", "06_experimentos_de_melhoria", "07_cenarios_e_resultados", "08_planejamentos_e_margens", "11_diagnostico_outubro", "09_figuras_e_tabelas", "10_revisao"]
for nome in ORDEM:
    print("Executando", nome)
    caderno = nbformat.read(f"{nome}.ipynb", as_version=4)
    NotebookClient(caderno, timeout=None, kernel_name="python3", resources={"metadata": {"path": "."}}).execute()
    nbformat.write(caderno, f"{nome}.ipynb")
print("Pipeline concluido")
```
